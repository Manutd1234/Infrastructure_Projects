# 02 — 13F sector rotation exploration

Inspect the latest 13F filings sector weights and look for crowded trades.

```python
import pandas as pd
from pathlib import Path

OUT = Path('../data/pipelines/thirteen_f_filings/outputs')
weights = pd.read_csv(OUT / 'sector_weights.csv')
holdings = pd.read_csv(OUT / 'holdings_with_sectors.csv')
weights.head()
```

## Latest quarter: which sectors are most crowded across all funds?

```python
latest = weights['quarter'].max()
latest_q = weights[weights['quarter'] == latest]
agg = latest_q.groupby('sector')['weight'].mean().sort_values(ascending=False)
print(f'Latest quarter: {latest}')
agg.head(12)
```

## Per-fund concentration (Herfindahl of sector weights)

```python
def hhi(s):
    return (s ** 2).sum()

latest_q.groupby('fund')['weight'].apply(hhi).sort_values(ascending=False)
```
