# 03 — Congress consensus and committee alignment

Explore the congress trading feed, consensus signals, and committee alignment.

```python
import pandas as pd
from pathlib import Path

OUT = Path('../CongressTrades/outputs')
trades = pd.read_csv(OUT / 'trades_with_sectors.csv')
consensus = pd.read_csv(OUT / 'ticker_consensus.csv')
committees = pd.read_csv(OUT / 'committee_summary.csv')
trades.head()
```

## What fraction of trades are committee-aligned?

```python
aligned = trades['committee_aligned'].mean()
print(f'Committee-aligned share: {aligned:.1%}')
trades.groupby('sector')['committee_aligned'].mean().sort_values(ascending=False)
```

## Top buys and sells by net signed USD

```python
print('Top BUYS:')
print(consensus.nlargest(10, 'net_signed_usd')[['ticker','issuer','net_signed_usd','n_buy','n_sell']])
print('\nTop SELLS:')
print(consensus.nsmallest(10, 'net_signed_usd')[['ticker','issuer','net_signed_usd','n_buy','n_sell']])
```
